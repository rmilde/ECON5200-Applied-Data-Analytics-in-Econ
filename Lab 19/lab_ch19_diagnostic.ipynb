````python
# -----------------------------------------------------------
# GUIDED — Run as-is
# Step 1: Import libraries and load data
# -----------------------------------------------------------
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.inspection import permutation_importance

RANDOM_STATE = 42
data = fetch_california_housing()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE)
````

## Part 3: Hyperparameter Tuning + XGBoost Comparison (10 min)

Tune the RF, then compare against XGBoost (gradient boosting).

````python
# -----------------------------------------------------------
# 🤖 AI EXPANSION — Co-Pilot required
# Copy the P.R.I.M.E. prompt above into Claude, then paste
# the generated code here. Run it and verify.
# -----------------------------------------------------------

# [Prep] Act as an expert Python Data Scientist specializing
# in SHAP explanations, interactive visualizations, and
# scikit-learn production workflows.
#
# [Request] I just completed a diagnosis-first lab where I
# compared Decision Trees, Ridge, Random Forests, and Gradient
# Boosting on California Housing data. I fixed evaluation bugs,
# diagnosed causal overclaiming from MDI, tuned hyperparameters
# with GridSearchCV, and generated SHAP waterfall + beeswarm
# plots. Now I need TWO artifacts:
#
# 1. A reusable `src/shap_utils.py` module with three functions:
#    - explain_prediction(model, X, idx) -> SHAP waterfall
#    - global_importance(model, X) -> SHAP beeswarm
#    - compare_importance(model, X, y) -> MDI vs SHAP side-by-side
#    Include type hints, docstrings, and error handling.
#
# 2. An interactive Plotly dashboard (or Streamlit app) with
#    ipywidgets sliders for n_estimators (1-500) and max_features
#    (1-8). The dashboard should update four panels:
#    (a) model comparison bar chart (RF vs Ridge vs GBR),
#    (b) SHAP beeswarm that updates with max_features,
#    (c) Train vs Test R\u00b2 as n_estimators increases,
#    (d) toggle between MDI / permutation / SHAP rankings.
#
# [Iterate] Use plotly.graph_objects, ipywidgets, shap, numpy,
# sklearn. Use the same variable names: X_train, X_test,
# y_train, y_test, data.feature_names. Do not use deprecated
# Plotly or SHAP functions.
#
# [Mechanism Check] Add inline comments explaining:
#   - How TreeExplainer differs from KernelExplainer
#   - Why SHAP values are additive (Shapley property)
#   - How ipywidgets observers trigger plot updates
#   - Why we re-fit inside the callback
#
# [Evaluate] Explain what the dashboard reveals about:
#   - The relationship between n_estimators, max_features,
#     and test performance
#   - Where MDI and SHAP rankings diverge and why
#   - The marginal value of additional trees beyond ~200

# PASTE AI-GENERATED CODE BELOW:

"""
Utility functions for SHAP-based model explainability.

Designed for tree-based sklearn models and compatible with
RandomForestRegressor, GradientBoostingRegressor, etc.
"""

from typing import Tuple
import numpy as np
import shap
import matplotlib.pyplot as plt
from sklearn.inspection import permutation_importance


def _validate_inputs(model, X):
    """Basic validation to ensure model is fitted and X is valid."""
    if model is None:
        raise ValueError("Model must not be None.")
    if X is None or len(X) == 0:
        raise ValueError("Input X must not be empty.")
    if not hasattr(model, "predict"):
        raise TypeError("Model must implement predict().")


def explain_prediction(model, X: np.ndarray, idx: int) -> None:
    """
    Generate a SHAP waterfall plot explaining a single prediction.

    Parameters
    ----------
    model : fitted sklearn model
    X : np.ndarray or DataFrame
        Feature matrix
    idx : int
        Index of observation to explain

    Notes
    -----
    - Uses TreeExplainer for efficiency (exact for tree models).
    - SHAP values are additive: prediction = base_value + sum(contributions).
    """
    _validate_inputs(model, X)

    if idx >= len(X):
        raise IndexError(f"Index {idx} out of bounds for X with length {len(X)}")

    # TreeExplainer leverages model structure → fast & exact for trees
    # KernelExplainer would approximate via sampling (model-agnostic, slower)
    explainer = shap.TreeExplainer(model)

    shap_values = explainer(X)

    # SHAP additivity (Shapley property):
    # prediction = expected_value + sum(feature contributions)
    shap.plots.waterfall(shap_values[idx])


def global_importance(model, X: np.ndarray) -> None:
    """
    Generate a SHAP beeswarm plot for global feature importance.

    Parameters
    ----------
    model : fitted sklearn model
    X : np.ndarray or DataFrame
        Feature matrix

    Notes
    -----
    - Beeswarm shows distribution of SHAP values per feature.
    - Captures magnitude + direction (unlike MDI).
    """
    _validate_inputs(model, X)

    explainer = shap.TreeExplainer(model)
    shap_values = explainer(X)

    shap.plots.beeswarm(shap_values)


def compare_importance(model, X: np.ndarray, y: np.ndarray) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Compare MDI (Mean Decrease Impurity) vs SHAP vs permutation importance.

    Parameters
    ----------
    model : fitted sklearn model
    X : np.ndarray
    y : np.ndarray

    Returns
    -------
    mdi : np.ndarray
    shap_imp : np.ndarray
    perm_imp : np.ndarray

    Notes
    -----
    - MDI is biased toward high-cardinality features.
    - SHAP is consistent and local-to-global aggregated.
    - Permutation importance measures performance degradation.
    """
    _validate_inputs(model, X)

    # MDI (built-in for tree models)
    if not hasattr(model, "feature_importances_"):
        raise AttributeError("Model does not support MDI feature_importances_.")

    mdi = model.feature_importances_

    # SHAP importance (mean absolute value)
    explainer = shap.TreeExplainer(model)
    shap_values = explainer(X)
    shap_imp = np.abs(shap_values.values).mean(axis=0)

    # Permutation importance (model-agnostic)
    perm = permutation_importance(model, X, y, n_repeats=5, random_state=42)
    perm_imp = perm.importances_mean

    # Plot side-by-side
    fig, ax = plt.subplots(figsize=(10, 6))
    x = np.arange(len(mdi))

    ax.bar(x - 0.25, mdi, width=0.25, label="MDI")
    ax.bar(x, shap_imp, width=0.25, label="SHAP")
    ax.bar(x + 0.25, perm_imp, width=0.25, label="Permutation")

    ax.set_title("Feature Importance Comparison")
    ax.legend()
    plt.show()

    return mdi, shap_imp, perm_imp
````

````python
import numpy as np
import plotly.graph_objects as go
from ipywidgets import interact, IntSlider, ToggleButtons
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score
from sklearn.inspection import permutation_importance
import shap

# Global storage for performance tracking
train_scores = []
test_scores = []
estimators_range = []


def build_dashboard(X_train, X_test, y_train, y_test, feature_names):

    def update(n_estimators, max_features, importance_type):

        # IMPORTANT:
        # We re-fit inside the callback because changing hyperparameters
        # changes the model structure → SHAP values and performance must be recomputed
        rf = RandomForestRegressor(
            n_estimators=n_estimators,
            max_features=max_features,
            random_state=42,
            n_jobs=-1
        )
        rf.fit(X_train, y_train)

        gbr = GradientBoostingRegressor(n_estimators=n_estimators, random_state=42)
        gbr.fit(X_train, y_train)

        ridge = Ridge()
        ridge.fit(X_train, y_train)

        # Predictions
        rf_train = r2_score(y_train, rf.predict(X_train))
        rf_test = r2_score(y_test, rf.predict(X_test))

        gbr_test = r2_score(y_test, gbr.predict(X_test))
        ridge_test = r2_score(y_test, ridge.predict(X_test))

        # Track learning curve
        estimators_range.append(n_estimators)
        train_scores.append(rf_train)
        test_scores.append(rf_test)

        # Panel (a): Model comparison
        fig1 = go.Figure(data=[
            go.Bar(name="RF", x=["Model"], y=[rf_test]),
            go.Bar(name="GBR", x=["Model"], y=[gbr_test]),
            go.Bar(name="Ridge", x=["Model"], y=[ridge_test])
        ])
        fig1.update_layout(title="Model Comparison (Test R²)")

        # Panel (b): SHAP beeswarm (approx via summary plot values)
        explainer = shap.TreeExplainer(rf)
        shap_values = explainer(X_train)

        shap_imp = np.abs(shap_values.values).mean(axis=0)

        fig2 = go.Figure()
        fig2.add_trace(go.Bar(
            x=shap_imp,
            y=feature_names,
            orientation='h'
        ))
        fig2.update_layout(title="SHAP Importance (updates with max_features)")

        # Panel (c): Learning curve
        fig3 = go.Figure()
        fig3.add_trace(go.Scatter(x=estimators_range, y=train_scores, name="Train R²"))
        fig3.add_trace(go.Scatter(x=estimators_range, y=test_scores, name="Test R²"))
        fig3.update_layout(title="Train vs Test R² vs n_estimators")

        # Panel (d): Importance comparison toggle
        if importance_type == "MDI":
            importance = rf.feature_importances_
        elif importance_type == "Permutation":
            perm = permutation_importance(rf, X_test, y_test, n_repeats=3)
            importance = perm.importances_mean
        else:  # SHAP
            importance = shap_imp

        fig4 = go.Figure()
        fig4.add_trace(go.Bar(
            x=importance,
            y=feature_names,
            orientation='h'
        ))
        fig4.update_layout(title=f"{importance_type} Importance")

        fig1.show()
        fig2.show()
        fig3.show()
        fig4.show()

    # ipywidgets uses observers:
    # when slider value changes → triggers update() automatically
    interact(
        update,
        n_estimators=IntSlider(min=1, max=500, step=10, value=100),
        max_features=IntSlider(min=1, max=8, step=1, value=4),
        importance_type=ToggleButtons(options=["MDI", "Permutation", "SHAP"])
    )
````
